```python
# now we will do the train test split for which data to go  for the traning purpose and which to go for the testing purpose okay 
from sklearn.model_selection import train_test_split 

x=df.drop(columns=["Churn"])
y=df["Churn"]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
# now we are building the pipleine for the columns for 1) Ordinal encoding 2)one hot encoding 3)standard scaling 
ordinal_features_encoding=["Subscription Type","Contract Length"]
one_hot_encoding_features=["Gender"]
standard_scaling_features=["Tenure","Usage Frequency","Support Calls","Payment Delay","Total Spend","Age"]




```

```python
# now we will build the separate pipline for the separate columns 
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OrdinalEncoder,OneHotEncoder


ordinal_encoding_pipeline=Pipeline(steps=[("encoder",OrdinalEncoder(categories=[["Basic","Standard","Premium"],["Monthly","Quarterly","Annual"]]))])

one_hot_encoding_pipeline=Pipeline(steps=[("one_hot_encoder",OneHotEncoder(handle_unknown='ignore'))])

standard_scaling_pipeline=Pipeline(steps=[("scaler",StandardScaler())])
```

```python
# now its time to collectively make the all independnty pipline into single transforming block 
#for this we will use column transforrmer okay

from sklearn.compose import ColumnTransformer 
column_transformer=ColumnTransformer(transformers=[
    ("standard_scaling_pipe",standard_scaling_pipeline,standard_scaling_features),
    ("ordinal_encoding_pipe",ordinal_encoding_pipeline,ordinal_features_encoding),
    ("one_hot_encoding_pipe",one_hot_encoding_pipeline,one_hot_encoding_features)]
)
```

```python
# now its time for training the model with logistic regression algorithm 
from sklearn.linear_model import LogisticRegression
logistic_regression_pipeline=Pipeline(steps=[("preprocessing",column_transformer),
                                    ("model",LogisticRegression())])
```

```python
# training the model 
logistic_regression_pipeline.fit(x_train,y_train)
```

# perdiction of the model

```python
prediction_data=logistic_regression_pipeline.predict(x_test)
```

# model evaluation starts 

```python
# now we will do the evaluation with the help of the confusuon matrix , precision, recall, f1 score , accuracy 
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report


# let us first make the confusion_matrix 
print(confusion_matrix(y_test,prediction_data))
print("the accuracy of the model is ")
print(accuracy_score(y_test,prediction_data))
print(classification_report(y_test,prediction_data))
```

```python
# now we will also evaluate the roc curve and roc_auc score to check which model is perfect in the prediction and overcame the type 1 and type 2 error 
from sklearn.metrics import roc_auc_score,roc_curve

probaboility_scores_of_xtest=logistic_regression_pipeline.predict_proba(x_test)[:,1]
```

```python
probaboility_scores_of_xtest
```

```python
fpr, tpr, thresholds =roc_curve(y_test,probaboility_scores_of_xtest)
```

```python
thresholds
```

```python
plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, color='blue', label=f'ROC curve (AUC = {auc_score:.3f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', label='Random guess')

# mark and label a few thresholds on the curve
for t in [0.2, 0.3, 0.5, 0.7, 0.9]:
    idx = np.argmin(np.abs(thresholds[1:] - t)) + 1
    plt.scatter(fpr[idx], tpr[idx], color='red', zorder=5)
    plt.annotate(f'{t}', (fpr[idx], tpr[idx]),
                 textcoords='offset points', xytext=(8, -12))

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate (Recall)')
plt.title('ROC Curve with Thresholds')
plt.legend(loc='lower right')
plt.grid(alpha=0.3)
plt.show()
```

```python
# roc_aucscore 

auc_score = roc_auc_score(y_test, probaboility_scores_of_xtest)
print(auc_score)
```
